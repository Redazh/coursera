```python
import pandas as pd
df = pd.read_csv('assets/fb441e62df2d58994928907a91895ec62c2c42e6cd075c2700843b89.csv')
df.head()
```

```python
# In this code cell, transform the Data_Value column

df_min = df[df['Element'] == 'TMIN']
df_max = df[df['Element'] == 'TMAX']
```

## Step 2
In order to visualize the data we would plot the min and max data for each day of the year between the years 2005 and 2014 across all weather stations. But we also need to find out when the min or max temperature in 2015 falls below the min or rises above the max for the previous decade.

If you did step 1 you have two Series objects with min and max times for the years 2005 through 2015. You can use Pandas `groupby` to create max and min temperature Series objects across all weather stations for each day of these years, and you can deal with the records for February 29 (the leap year) by dropping them.

__hint: when I finished this step, I had two DataFrame objects, each with exactly 4015 observations in them__

```python
# create a DataFrame of maximum temperature by date
df_max['Date'] = pd.to_datetime(df_max['Date'])
df_max['Month'] = pd.DatetimeIndex(df_max['Date']).month
df_max['Day'] = pd.DatetimeIndex(df_max['Date']).day
df_max['Year'] = pd.DatetimeIndex(df_max['Date']).year
grouped_max = df_max.groupby(['Month','Day','Year']).max()
grouped_max = grouped_max.reset_index()
grouped_max = grouped_max.drop(grouped_max[(grouped_max['Day'] == 29) & (grouped_max['Month'] == 2)].index)
# create a DataFrame of minimum temperatures by date
df_min['Date'] = pd.to_datetime(df_min['Date'])
df_min['Month'] = pd.DatetimeIndex(df_min['Date']).month
df_min['Day'] = pd.DatetimeIndex(df_min['Date']).day
df_min['Year'] = pd.DatetimeIndex(df_min['Date']).year
grouped_min = df_min.groupby(['Month','Day','Year']).min()
grouped_min = grouped_min.reset_index()
grouped_min = grouped_min.drop(grouped_min[(grouped_min['Day'] == 29) & (grouped_min['Month'] == 2)].index)
```

## Step 3
Now that you have grouped the daily max and min temperatures for each day of the years 2005 through 2015, you can separate out the data for 2015. Then you can use the Pandas `groupby` function to find the max and min of the temperature data for each __day of the year__ for the 2005-2014 data.

__hint: at the end of this step I had two DataFrames, one of maximum and the other of minimum values, which each had 365 observations in them. I also had another pair of similar DataFrames but only for the year 2015.__

```python
# calculate the minimum and maximum values for the day of the year for 2005 through 2014
df_2005_2014_max = grouped_max[grouped_max['Year'].between(2005, 2014)]
df_2005_2014_min = grouped_min[grouped_min['Year'].between(2005, 2014)]
max_values_2005_2014 = df_2005_2014_max.groupby(['Month', 'Day'])['Data_Value'].max()
min_values_2005_2014 = df_2005_2014_min.groupby(['Month', 'Day'])['Data_Value'].min()
# calculate the minimum and maximum values for the years 2015
# Créez un DataFrame pour l'année 2015
df_2015_max = grouped_max[grouped_max['Year'] == 2015]
df_2015_min = grouped_min[grouped_min['Year'] == 2015]
# Calculez les valeurs maximales et minimales pour chaque jour de l'année pour 2015
max_values_2015 = df_2015_max.groupby(['Month', 'Day'])['Data_Value'].max()
min_values_2015 = df_2015_min.groupby(['Month', 'Day'])['Data_Value'].min()
```

## Step 4
Now it's time to plot! You need to explore matplotlib in order to plot line graphs of the min and max temperatures for the years 2005 through 2014 and to scatter plot __only__ the daily 2015 temperatures that exceeded those values.

```python
import matplotlib.pyplot as plt
from calendar import month_abbr

# put your plotting code here!
# Convert les abréviations des mois en noms complets pour l'étiquetage des graphiques
month_names = month_abbr[1:]
# Créez des listes pour les jours de l'année (1 à 365)
day_of_year = list(range(1, 366))
# Créez des listes pour les valeurs maximales et minimales de 2005 à 2014
max_values_2005_2014_list = list(max_values_2005_2014)
min_values_2005_2014_list = list(min_values_2005_2014)
# Créez des listes pour les valeurs maximales et minimales de 2015
max_values_2015_list = list(max_values_2015)
min_values_2015_list = list(min_values_2015)
# Créez une liste pour les jours de 2015 où la température a dépassé les valeurs de 2005-2014
days_exceeding_max = [i for i, (value2014, value2015) in enumerate(zip(max_values_2005_2014_list, max_values_2015_list)) if value2015 > value2014]
days_exceeding_min = [i for i, (value2014, value2015) in enumerate(zip(min_values_2005_2014_list, min_values_2015_list)) if value2015 < value2014]
# Créez un graphique linéaire des températures maximales et minimales de 2005-2014
plt.figure(figsize=(10, 6))
plt.plot(day_of_year, max_values_2005_2014_list, color='red', label='Max Temp (2005-2014)')
plt.plot(day_of_year, min_values_2005_2014_list, color='blue', label='Min Temp (2005-2014)')
# Créez un nuage de points pour les jours de 2015 où la température a dépassé les valeurs de 2005-2014
plt.scatter(days_exceeding_max, [max_values_2015_list[i] for i in days_exceeding_max], color='green', label='Record High (2015)')
plt.scatter(days_exceeding_min, [min_values_2015_list[i] for i in days_exceeding_min], color='purple', label='Record Low (2015)')
# Étiquetez les axes et ajoutez une légende
plt.xlabel('days of year')
plt.ylabel('Temperature')
plt.title('Low & High Temperature between 2005-2014 \n and Record Braking Temperatures of 2015')
plt.legend(loc='best')
# Personnalisez l'axe des x pour afficher les mois
plt.xticks([1, 32, 60, 91, 121, 152, 182, 213, 244, 274, 305, 335],
           month_names, rotation=45)
# Affichez la grille
plt.grid(True, linestyle='--', alpha=0.7)
plt.gca().fill_between(day_of_year, 
                       max_values_2005_2014_list, min_values_2005_2014_list, 
                       facecolor='gray', 
                       alpha=0.25)
# Affichez le graphique
plt.show()
```
